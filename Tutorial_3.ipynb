#### Read Data

* The main object in pandas is called a DataFrame, which is similar to a table with rows and columns.

The following cell of code loads data from a CSV file into a Pandas DataFrame.

note:
* pd.read_csv(...) reads a CSV file using the Pandas library.
* 'data/diamonds.csv' is the location of the file (change it to the location of your CSV file)
* sep=',' tells Pandas that the values in the file are separated by commas.
* index_col=0 tells Pandas to use the first column in the CSV file as the row index instead of treating it as a regular data column.
* diamonds_df = saves the loaded table in a variable called diamonds_df.

After running it, diamonds_df contains the diamonds dataset and can be used for data analysis.

```python
diamonds_df = pd.read_csv('data/diamonds.csv', sep=',' , index_col=0)
```

* In The following cell, The head() method displays the first rows of a DataFrame.
* Recall that diamonds_df is the DataFrame that contains the diamonds dataset.
* .head(10) tells Pandas to display the first 10 rows of the DataFrame (chane it to display more or fewer rows using a different number).
* This method is commonly used to quickly inspect the data after loading it and verify that it looks correct.

```python
diamonds_df.head(10)
```

### choose columns, filter rows & sort values
There are several ways to choose a subset of columns from a DataFrame.
* To show only one chosen column out of diamonds_df use the name of it inside square brackets (or a list of names to show more than one column).

```python
diamonds_df["cut"]
```

```python
diamonds_df[["color", "price"]]
```

* Alternatively, use loc attribute

```python
diamonds_df.loc[:,"cut"]
```

```python
diamonds_df.loc[:,["color","price"]]
```

* You can also choose row with loc attribute, for example chose the 4'th row (note it will be displayed as transpose):

```python
diamonds_df.loc[4]
```

* You  can also choose rows 4 - 8

```python
diamonds_df.loc[4:8]
```

* or, choose rows by condition:

```python
diamonds_df.loc[diamonds_df['depth'] < 61]
```

```python
# note the boolean output inside the []
diamonds_df['depth'] < 61
```

* You can sort rows by chosen column using sort values method (note the difference between inplace=False/True and .reset_index())

```python
diamonds_df.sort_values(by='carat', ascending=False, inplace=False)
```

```python
diamonds_df.sort_values(by='carat', ascending=False, inplace=True)
```

```python
diamonds_df.reset_index(inplace=True)
```

### Create new colum in the data frame

```python
diamonds_df["price_per_carat"] = diamonds_df["price"] / diamonds_df["carat"]

diamonds_df["is_expensive"] = diamonds_df["price"] > 10000

diamonds_df
```

```python
diamonds_df.shape[0]
```

```python
diamonds_df.shape[0] - diamonds_df['is_expensive'].sum()
```

```python
diamonds_df["rate"] = np.where(diamonds_df["price"] > 10000, 'expensive', 'cheap')
diamonds_df.head()
```
